```python
#Finding the distance between two points

import math

x1 = float(input(" Enter x1:"))
x2 = float(input(" Enter x2:"))
y1 = float(input(" Enter y1:"))
y2 = float(input(" Enter y2:"))


distance = math.pow((x2 - x1),math.sqrt, 2) + math.pow((y2 - y1),math.sqrt, 2)

print(math.floor(distance))

#When using a library, it's more practical because it is faster and easier and helps finding with errors.
```
